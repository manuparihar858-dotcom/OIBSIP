This project analyzes retail sales data to understand sales trends,
customer characteristics, and product performance using Python.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Data.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df_clean = df.copy()
```

```python
df_clean[df_clean.isnull().any(axis=1)]
```

```python
df_clean = df_clean.dropna(
    subset=['quantiy', 'price_per_unit', 'cogs', 'total_sale']
)
```

```python
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())
```

```python
df_clean.isnull().sum()
```

```python
df_clean.shape
```

```python
df_clean['sale_date'] = pd.to_datetime(df_clean['sale_date'])
```

```python
df_clean.info()
```

```python
df_clean['sale_date'] = pd.to_datetime(df_clean['sale_date'])
```

```python
df_clean['sale_date'].dtype
```

```python
df_clean['year'] = df_clean['sale_date'].dt.year
df_clean['month'] = df_clean['sale_date'].dt.month
df_clean['month_name'] = df_clean['sale_date'].dt.month_name()
df_clean['quarter'] = df_clean['sale_date'].dt.quarter
```

```python
df_clean[['sale_date', 'year', 'month', 'month_name', 'quarter']].head()
```

```python
df_clean.describe()
```

```python
print("Mean:")
print(df_clean[['age', 'quantiy', 'price_per_unit', 'cogs', 'total_sale']].mean())

print("\nMedian:")
print(df_clean[['age', 'quantiy', 'price_per_unit', 'cogs', 'total_sale']].median())

print("\nMode:")
print(df_clean[['age', 'quantiy', 'price_per_unit', 'cogs', 'total_sale']].mode().iloc[0])

print("\nStandard Deviation:")
print(df_clean[['age', 'quantiy', 'price_per_unit', 'cogs', 'total_sale']].std())
```

```python
monthly_sales = df_clean.groupby(
    ['year', 'month', 'month_name'],
    as_index=False
)['total_sale'].sum()
```

```python
monthly_sales = monthly_sales.sort_values(['year', 'month'])
```

```python
monthly_sales
```

```python
highest_month = monthly_sales.loc[monthly_sales['total_sale'].idxmax()]
lowest_month = monthly_sales.loc[monthly_sales['total_sale'].idxmin()]

print("Highest Sales Month:")
print(highest_month)

print("\nLowest Sales Month:")
print(lowest_month)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_sales,
    x='month_name',
    y='total_sale',
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_clean['year_month'] = df_clean['sale_date'].dt.to_period('M').astype(str)
```

```python
monthly_sales = (
    df_clean.groupby('year_month', as_index=False)['total_sale']
    .sum()
)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_sales,
    x='year_month',
    y='total_sale',
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Monthly Sales Trend

Monthly sales were calculated by grouping transactions according to their year and month.

The resulting trend shows how sales changed over time and helps identify periods with relatively higher or lower sales activity.

This information can support inventory planning, promotional timing, and future sales forecasting.

```python
quarterly_sales = (
    df_clean.groupby(['year', 'quarter'], as_index=False)['total_sale']
    .sum()
)

quarterly_sales
```

```python
quarterly_sales['quarter_label'] = (
    quarterly_sales['year'].astype(str)
    + ' Q'
    + quarterly_sales['quarter'].astype(str)
)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=quarterly_sales,
    x='quarter_label',
    y='total_sale'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

The monthly sales trend shows variation in sales performance across the observed months.

The changes in monthly revenue indicate that sales activity was not consistent throughout the period. Identifying stronger and weaker months can help the business plan inventory and promotional activities more effectively.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_clean,
    x='age',
    bins=15,
    kde=True
)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

### Observation

The age distribution shows the range and concentration of customers in the dataset.
This helps identify the major customer age groups and understand which age ranges
are most represented in the retail sales data.

```python
gender_sales = (
    df_clean.groupby('gender', as_index=False)['total_sale']
    .sum()
)

gender_sales
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=gender_sales,
    x='gender',
    y='total_sale'
)

plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()
```

### Observation

The gender-wise sales analysis compares the total sales generated by different
customer genders. This provides insight into the contribution of each gender
segment to overall sales.

```python
top_products = (
    df_clean.groupby('category', as_index=False)['total_sale']
    .sum()
    .sort_values('total_sale', ascending=False)
)

top_products
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_products,
    x='category',
    y='total_sale'
)

plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

Revenue varies across the different product categories.
The category with the highest total revenue represents the largest
contributor to overall sales, while lower-revenue categories may require
further analysis to understand their performance.

```python
numeric_cols = [
    'age',
    'quantiy',
    'price_per_unit',
    'cogs',
    'total_sale'
]

correlation = df_clean[numeric_cols].corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

### Observation

The under 20 age group recorded the highest average transaction value, at approximately 559.03.

This result suggests that younger customers represented a relatively high average transaction value in this dataset. Further analysis with a larger dataset would be useful to determine whether this pattern remains consistent over time.

```python
df_clean['age_group'] = pd.cut(
    df_clean['age'],
    bins=[0, 20, 30, 40, 50, 60, 100],
    labels=['<20', '20-29', '30-39', '40-49', '50-59', '60+']
)

age_group_sales = (
    df_clean.groupby('age_group', observed=True)['total_sale']
    .mean()
    .reset_index()
)

age_group_sales
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=age_group_sales,
    x='age_group',
    y='total_sale'
)

plt.title('Average Transaction Value by Customer Age Group')
plt.xlabel('Age Group')
plt.ylabel('Average Transaction Value')
plt.tight_layout()
plt.show()
```

### Observation

The average transaction value differs across customer age groups.
The age group with the highest average transaction value represents
customers who tend to make higher-value purchases per transaction.
This can help identify customer segments that may respond differently
to targeted promotions and product strategies.

```python
print("Total Revenue:", df_clean['total_sale'].sum())
print("Average Transaction:", df_clean['total_sale'].mean())
print("Median Transaction:", df_clean['total_sale'].median())

print("\nTop Category:")
print(top_products.iloc[0])

print("\nHighest Sales Quarter:")
print(quarterly_sales.loc[quarterly_sales['total_sale'].idxmax()])

print("\nHighest Average Transaction Age Group:")
print(age_group_sales.loc[age_group_sales['total_sale'].idxmax()])
```

## Conclusion

The exploratory analysis provided an overview of sales performance, customer characteristics, and category level revenue in the retail sales dataset.

After cleaning, the dataset contained 1997 records with a total revenue of 911720. The average transaction value was 456.54, while the median transaction value was 150. The difference between these values indicates that transaction values were not evenly distributed.

Electronics was the highest revenue generating category, with total sales of 313810. The fourth quarter of 2022 recorded the highest quarterly revenue, with sales of 210030. The under 20 age group recorded the highest average transaction value, at approximately 559.03.

Based on these findings, the business can focus on maintaining strong performance in the Electronics category, prepare inventory and promotional activities ahead of the stronger fourth quarter period, and explore targeted offers for younger customers while monitoring whether their higher average transaction value continues in future sales data.

Overall, the analysis demonstrates how basic exploratory data analysis can be used to identify sales patterns, understand customer segments, and support data driven business decisions.
